### **1.2 ZüriWieNeu reports**

#### **1.2.1 Loading the csv. ZüriWieNeu reports as a Pandas dataframe and inspect the data and metadata.**

```python
import pandas as pd
import geopandas as gpd

# 1. Load the csv
zwn_df = pd.read_csv("data/raw/stzh.zwn_meldungen_p.csv")

print("Data loaded successfully!")

# 2. Peek into the data
display(zwn_df.head(4))
display(zwn_df.tail(4))

# 3. View shape & columns of the pd dataframe
print(zwn_df.shape)
print(zwn_df.columns.values)
display(zwn_df.info())
```

#### **1.2.2 Conversion of datetime data to Pandas datetime objects.**

```python
# Conversion to Pandas datetime objects

for col in zwn_df.columns:
    if col.endswith("_datetime"): 
        zwn_df[col] = pd.to_datetime(zwn_df[col], errors="coerce") # if someting is not recognised it is transformed in to NaT by errors="coerce"

display(zwn_df.info())
```

#### **1.2.3 Filling in missing dates in the column agency_sent_datetime with an interpolation.**

```python
# 1. Make a safe copy
zwn_interp_df = zwn_df.copy()

# 2. Use interpolate() between the day before and day after to calculate the missing values in the column agency_sent_datetime
zwn_interp_df["agency_sent_datetime"] = zwn_interp_df["agency_sent_datetime"].interpolate()

display(zwn_interp_df.info())
```

#### **1.2.4 Normalize the datetime objects becaus only the date and not the exact time is relevant for the analysis.**

```python
#Normalization of the datetime objects and save it as new columns
for col in zwn_interp_df.columns:
    if col.endswith("_datetime"):
        new_col = col.replace("_datetime", "_date")
        zwn_interp_df[new_col] = (zwn_interp_df[col].dt.normalize())

display(zwn_interp_df["requested_date"].min())
```

#### **1.2.5 Rename the geometry column of the zwn_interp_df.**

```python
# Rename geometry column
zwn_interp_df = zwn_interp_df.rename(columns={"geometry": "geometry_zwn"})

# View change
display(zwn_interp_df.info())
```

#### **1.2.6 Save the zwn_interp_df of the Züri wie neu reports to a new csv file.**

```python
# Use index=False to prevent saving the sequential row numbers of the Index in the csv
zwn_interp_df.to_csv("data/processed/stzh.zwn_meldungen_p_datetime64_normalized.csv", index=False)

print("File successfully saved!")
```

#### **1.2.7 Create a geo dataframe of the Züri wie neu reports.**

```python
from shapely import wkt

# WKT -> Convert geometry
zwn_interp_df["geometry_zwn"] = zwn_interp_df["geometry_zwn"].apply(wkt.loads)

# Create a GeoDataFrame
zwn_interp_gdf = gpd.GeoDataFrame(zwn_interp_df, geometry="geometry_zwn")

# Set CRS (LV95 / EPSG:2056)
zwn_interp_gdf = zwn_interp_gdf.set_crs(epsg=2056)

# 2. Peek into the data an inspect the coordinate reference system
print(zwn_interp_gdf.crs)
display(zwn_interp_gdf.head(4))

# 3. View shape & columns of the geo dataframe
display(zwn_interp_gdf.info())
```

#### **1.2.8 Save the geo dataframe of the Züri wie neu reports to a new gpkg file.**

```python
# Export as a GeoPackage (for sharing with QGIS/ArcGIS users)
zwn_interp_gdf.to_file("data/processed/stzh.zwn_meldungen_p_processed.gpkg")

print("File successfully saved!")
```

### **1.3 Population data of the City of Zurich**

#### **1.3.1 Loading the xlsx. population date of the city of Zurich  as a Pandas dataframe and inspect the data and metadata.**

```python
# 1. Load the xlsx
file_path = "data/raw/Bevoelkerung_nach-Herkunft-Stadtkreis-Stadtquartier.xlsx"

# Creat a list with the years of the sheets (years are the sheets of the excel)
years = [str(year) for year in range(2025, 2013, -1)]

# An empty list to collect the DataFrames of the diffrent year sheets of the excel
dfs = []

for year in years:

    # Import Excel
    df = pd.read_excel(
        file_path,
        sheet_name=year,
        usecols="A:B",
        skiprows=8,
        header=0,
        engine="openpyxl")

    # Rename columns
    df.columns = ["qname", f"pop_tot_{year}"]

    # Remove blank lines
    df = df.dropna(subset=["qname"])

    dfs.append(df)

# Consolidate districts across all years
zh_pop_df = dfs[0]

for df in dfs[1:]:
    zh_pop_df = zh_pop_df.merge(df, on="qname", how="outer")

print("Data loaded successfully!")

# 2. Peek into the data
display(zh_pop_df.head(4))
display(zh_pop_df.tail(4))

# 3. View shape & columns of the pd dataframe

display(zh_pop_df.info())
```

#### **1.3.2 Inspect the districts (qname) column.**

```python
zh_pop_df["qname"]
```

#### **1.3.3 Remove the rows Kreis and  Ganze Stadt from the dataframe as this rows are not needed for the anlaysis.**

```python
zh_pop_dist_df = zh_pop_df.copy()

zh_pop_dist_df = zh_pop_dist_df[
    (~zh_pop_dist_df["qname"].str.startswith("Kreis")) &
    (zh_pop_dist_df["qname"] != "Ganze Stadt")]

zh_pop_dist_df = zh_pop_dist_df.reset_index(drop=True)

# 2. Peek into the data
display(zh_pop_dist_df.head(4))

# 3. View shape & columns of the pd dataframe

display(zh_pop_dist_df.info())

zh_pop_dist_df["qname"]
```

#### **1.3.4 Save the geo dataframe with population and districts of the city of Zurich to a new gpkg file.**

```python
# Use index=False to prevent saving the sequential row numbers
zh_pop_dist_df.to_csv("data/processed/Bevoelkerung_nach-Herkunft-Stadtkreis-Stadtquartier_urbandists.csv", index=False)

print("File successfully saved!")
```

### **1.4 Statistical districts data of the City of Zurich**

#### **1.4.1 Loading the csv. statistical districts in Zurich of the city of Zurich as a Pandas dataframe and inspect the data and metadata.**

```python
# 1. Load the csv
zh_dist_df = pd.read_csv("data/raw/stzh.adm_statistische_quartiere_map.csv")

print("Data loaded successfully!")

# 2. Explore the columns to find the geometry column 
print("Columns in CSV:", zh_dist_df.columns.values)

# 3. Peek into the data
display(zh_dist_df.head(4))

# 3. View shape & columns of the pd dataframe
print(zh_dist_df.shape)
print(zh_dist_df.columns.values)
display(zh_dist_df.info())
```

#### **1.4.2 Create a geo dataframe of of the district dataframe.**

```python
from shapely import wkt

# WKT -> Convert geometry
zh_dist_df["geometry"] = zh_dist_df["geometry"].apply(wkt.loads)

# Create a GeoDataFrame
zh_dist_gdf = gpd.GeoDataFrame(zh_dist_df, geometry="geometry")

# Set CRS (probably LV95 / EPSG:2056)
zh_dist_gdf = zh_dist_gdf.set_crs(epsg=2056)

# 2. Peek into the data and inspect the Coordinate Reference System
print(zh_dist_gdf.crs)
display(zh_dist_gdf.head(4))

# 3. View shape & columns of the geo dataframe
print(zh_dist_gdf.shape)
print(zh_dist_gdf.columns.values)
display(zh_dist_gdf.info())
```

#### **1.4.3 Plot the geometry to see if the geometry is usable.**

```python
# Instantly visualize the geometric shapes
zh_dist_gdf.plot(figsize=(8, 5), color="dodgerblue", edgecolor="black");
```

#### **1.4.4 Rename the geometry column.**

```python
# Rename geometry column
zh_dist_gdf = zh_dist_gdf.rename(columns={"geometry": "geometry_poly"})

# View change
display(zh_dist_gdf.info())
```

#### **1.4.5 Save geo dataframe of the districts of the city of Zurich to a new gpkg file.**

```python
# Export as a GeoPackage (For sharing with QGIS/ArcGIS users)
zh_dist_gdf.to_file("data/processed/stzh.adm_statistische_quartiere_map_processed.gpkg")

print("File successfully!")
```

### **1.5 Merge the statistical districts data of the city of Zurich with the population data of the city of Zurich**

#### **1.5.1 Merge the geo dataframe districts with the geo dataframe populations.**

```python
# Merge the two dataframes
zh_dist_pop_gdf = zh_dist_gdf.merge(
    zh_pop_dist_df,
    on="qname",
    how="outer"
)

print("Merge successfully!")

# Set the active geometry column
zh_dist_pop_gdf = zh_dist_pop_gdf.set_geometry("geometry_poly")

# Set CRS
zh_dist_pop_gdf = zh_dist_pop_gdf.set_crs(epsg=2056)

# Peek into the data
print(zh_dist_pop_gdf.crs)
display(zh_dist_pop_gdf.head(4))

# View shape & columns
print(zh_dist_pop_gdf.shape)
print(zh_dist_pop_gdf.columns.values)
zh_dist_pop_gdf.info()
```

#### **1.5.2 Calculate the  area for each district and add it as new column to the merged geo dataframe.**

```python
zh_dist_pop_gdf["area_m2"] = zh_dist_pop_gdf.geometry.area
zh_dist_pop_gdf["area_km2"] = (zh_dist_pop_gdf["area_m2"] / 1_000_000).round(2)

display(zh_dist_pop_gdf.head(4))
```

#### **1.5.3 Calculate the population denstiy of each district and year.**

```python
for year in range(2014, 2026):

    pop_col = f"pop_tot_{year}"
    density_col = f"pop_density_{year}"

    zh_dist_pop_gdf[density_col] = (zh_dist_pop_gdf[pop_col] / zh_dist_pop_gdf["area_km2"]).round(0)
    
display(zh_dist_pop_gdf.head(4))
```

#### **1.5.4 Calculate the population mean across all years of each district.**

```python
# Select all pop_tot columns
pop_tot_cols = [col for col in zh_dist_pop_gdf.columns if col.startswith('pop_tot_')]

# Calculate the average and add it as a new column
zh_dist_pop_gdf['pop_mean'] = zh_dist_pop_gdf[pop_tot_cols].mean(axis=1).round(0).astype(int)

zh_dist_pop_gdf[['qname', 'pop_mean'] + pop_tot_cols].head()

display(zh_dist_pop_gdf.info())
```

#### **1.5.5 Calculate the population mean denstiy of each district.**

```python
zh_dist_pop_gdf["pop_mean_density"] = (zh_dist_pop_gdf["pop_mean"] / zh_dist_pop_gdf["area_km2"]).round(0)
    
display(zh_dist_pop_gdf.head(4))
display(zh_dist_pop_gdf.info())
```

#### **1.5.6 Plot the geometry and the population mean to see if the geometry is usable.**

```python
# Instantly visualize the normalized density ratio
zh_dist_pop_gdf.plot(
    column="pop_mean",
    cmap="YlOrRd",
    figsize=(8, 5),
    legend=True,
    legend_kwds={"label": "Population Density (Inhabitants / Area)"},
    edgecolor="grey",);
```

#### **1.5.7 Save geo dataframe of the districts and population with the new columns to a new gpkg file.**

```python
# Export as a GeoPackage (For sharing with QGIS/ArcGIS users)
zh_dist_pop_gdf.to_file("data/processed/stzh.adm.pop_statistische_quartiere_map_processed.gpkg")

print("File successfully saved!")
```

### **1.6 Performe a nearest neighbour join**

#### **1.6.1 Performe a nearest neighbour join of the geo dataframe ZüriWieNeu and the merged disricts and populations**

```python
zwn_zh_dist_pop_gdf = gpd.sjoin_nearest(
    zwn_interp_gdf,
    zh_dist_pop_gdf,
    how="left",
    distance_col="dist_to_polygon")

print("Join successfully!")

pd.set_option("display.max_columns", None)

# 2. Peek into the data
print(zwn_zh_dist_pop_gdf.crs)
display(zwn_zh_dist_pop_gdf.head(4))

# 3. View shape & columns of the pd dataframe
print(zwn_zh_dist_pop_gdf.shape)
print(zwn_zh_dist_pop_gdf.columns.values)
display(zwn_zh_dist_pop_gdf.info())
```

#### **1.6.2 Check which geometry is active and plot the geometry to see if the geometry is usable.**

```python
print(zwn_zh_dist_pop_gdf.geometry.name)  # should show "geometry_zwn"

# Instantly visualize the geometric shapes
zwn_zh_dist_pop_gdf.plot(figsize=(8, 5), color="dodgerblue", edgecolor="black");
```

#### **1.6.3 Inspect if there are NaN values after join.**

```python
nan_rows = zwn_zh_dist_pop_gdf[zwn_zh_dist_pop_gdf["index_right"].isna()]

if not nan_rows.empty:
    print("NaN values found.")
else:
    print("No NaN values found.")

nan_rows
```

#### **1.6.4 Check if there are duplicates after the join.**

```python
duplicates = zwn_zh_dist_pop_gdf[zwn_zh_dist_pop_gdf.duplicated(subset="objectid_left", keep=False)]

if not duplicates.empty:
    print("Duplicate values found.")
else:
    print("No duplicate values found.")

display(duplicates)
```

#### **1.6.5 Map the duplicates to see where they are located.**

```python
import matplotlib.pyplot as plt

# Find points with multiple matches
duplicate_points = zwn_zh_dist_pop_gdf[zwn_zh_dist_pop_gdf.duplicated(subset="objectid_left",keep=False)]

fig, ax = plt.subplots(figsize=(10, 8))

# Plot neighborhood polygons
zh_dist_pop_gdf.plot(ax=ax, color="white", edgecolor="black")

duplicate_points.plot(ax=ax, color="red", markersize=40)

# Add neighborhood names
for idx, row in zh_dist_pop_gdf.iterrows():
    point = row["geometry_poly"].representative_point()

    ax.annotate(
        text=row["qname"],
        xy=(point.x, point.y),
        ha="center",
        fontsize=8,
        color="black")

ax.set_title("Points without a polygon assignment")
plt.show()
```

#### **1.6.6 Drop the duplicates because there are only two of them and this has no effect on the analysis.**

```python
zwn_zh_dist_pop_gdf = (zwn_zh_dist_pop_gdf.drop_duplicates(subset="objectid_left"))

print("Duplicates droped!")

display(zwn_zh_dist_pop_gdf.info())
```

#### **1.6.7 Check if the duplicates have been droped.**

```python
duplicates = zwn_zh_dist_pop_gdf[zwn_zh_dist_pop_gdf.duplicated(subset="objectid_left", keep=False)]

if not duplicates.empty:
    print("Duplicate values found.")
else:
    print("No duplicate values found.")

display(duplicates)
```

## **2. Spatial analysis**

### **2.1 Visualize the numbers of cases across the city over time.**

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Aggregate cases per month
cases_over_time = (
    zwn_zh_dist_pop_gdf
    .groupby(zwn_zh_dist_pop_gdf["requested_date"].dt.to_period("M"))
    .size()
    .reset_index(name="count")
)
cases_over_time["requested_date"] = cases_over_time["requested_date"].dt.to_timestamp()
cases_over_time = cases_over_time.sort_values("requested_date")

# COVID-19 pandemic
corona_start = pd.Timestamp("2020-03-01")
corona_end   = pd.Timestamp("2022-04-30")

# Plot
fig, ax = plt.subplots(figsize=(12, 5))

# Highlight the COVID-19 period
ax.axvspan(corona_start, corona_end,
           color="red", alpha=0.12, label="COVID-19 period", zorder=1)
ax.axvline(corona_start, color="red", linewidth=1, linestyle="--", alpha=0.6, zorder=2)
ax.axvline(corona_end,   color="red", linewidth=1, linestyle="--", alpha=0.6, zorder=2)

# Label COVID-19 period
y_max = cases_over_time["count"].max()
ax.text(
    corona_start + (corona_end - corona_start) / 2,
    y_max * 0.97,
    "COVID-19 period",
    ha="center", va="top", fontsize=9,
    color="red", fontweight="bold"
)

ax.plot(
    cases_over_time["requested_date"],
    cases_over_time["count"],
    color="blue",
    linewidth=2,
    marker="o",
    markersize=4,
    markerfacecolor="white",
    markeredgewidth=1.5,
    zorder=3,
    label="Number of cases"
)

# Format axes
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b %Y"))
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
plt.xticks(rotation=45, ha="right")
ax.set_xlabel("Date", fontsize=11)
ax.set_ylabel("Number of cases", fontsize=11)
ax.set_title("Cases over time", fontsize=14, fontweight="bold")
ax.legend(fontsize=9)
ax.grid(axis="y", linestyle="--", alpha=0.5)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
fig.savefig("outputs/zwn_cases_over_time.png", bbox_inches="tight")
plt.show()
```

### **2.2 Identify the number of cases per district.**

```python
cases_per_district = (zwn_zh_dist_pop_gdf.groupby("qname").size().reset_index(name="case_count"))
display(cases_per_district)
```

### **2.3 Merge the cases per district with the district polygon geo dateframe.**

```python
zh_dist_pop_gdf = zh_dist_pop_gdf.merge(cases_per_district, on="qname", how="left")

zh_dist_pop_gdf["case_count"] = zh_dist_pop_gdf["case_count"].fillna(0)

display(zh_dist_pop_gdf.info())

display(zh_dist_pop_gdf.head(4))
```

### **2.4 Visualize the numbers of cases per district to check if the merge was successful.**

```python
zh_dist_pop_gdf.sort_values(
    by="case_count",
    ascending=False
).plot(
    x="qname",
    y="case_count",
    kind="bar",
    color="blue",
    figsize=(12, 6)
)

zh_dist_pop_gdf.plot(
    column="case_count",
    legend=True,
    cmap="OrRd"
)
```

### **2.5 Merge the cases per district per year with the district polygon geo dateframe.**

```python
zwn_zh_dist_pop_gdf['year'] = zwn_zh_dist_pop_gdf['requested_date'].dt.year.astype('int64')

display(zwn_zh_dist_pop_gdf.info())


cases_per_district_year = (
    zwn_zh_dist_pop_gdf[zwn_zh_dist_pop_gdf['year'].between(2014, 2025)]
    .groupby(["qname", "year"])
    .size()
    .reset_index(name="case_count")
)

# Wide-format: one column per year
cases_wide = cases_per_district_year.pivot(
    index="qname",
    columns="year",
    values="case_count"
).fillna(0).astype(int).reset_index()

# Rename columns: 2014 -> case_count_2014
cases_wide.columns = [
    f"case_count_{col}" if isinstance(col, int) else col
    for col in cases_wide.columns
]
cases_wide.columns.name = None

# Merge with the GeoDataFrame
zh_dist_pop_gdf = zh_dist_pop_gdf.merge(cases_wide, on="qname", how="outer")

print("Merge erfolgreich!")
display(zh_dist_pop_gdf.info())
display(zh_dist_pop_gdf.head(4))
```

### **2.6 Visualize the cases per district in absolut numbers and per 1000 residents**

```python
import matplotlib.pyplot as plt

# Calculate cases per 1000 residents
zh_dist_pop_gdf["cases_per_1000"] = (
    zh_dist_pop_gdf["case_count"] / zh_dist_pop_gdf["pop_mean"]
) * 1000

fig, axes = plt.subplots(1, 2, figsize=(24, 12))

# Plot 1: Total Cases
vmin1 = zh_dist_pop_gdf["case_count"].min()
vmax1 = zh_dist_pop_gdf["case_count"].max()

zh_dist_pop_gdf.plot(
    column="case_count",
    cmap="OrRd",
    legend=True,
    vmin=vmin1,
    vmax=vmax1,
    edgecolor="black",
    linewidth=0.5,
    ax=axes[0]
)

for idx, row in zh_dist_pop_gdf.iterrows():
    point = row["geometry_poly"].representative_point()
    axes[0].annotate(
        text=row["qname"],
        xy=(point.x, point.y),
        ha="center", va="center",
        fontsize=8, color="black"
    )

axes[0].set_title("Total cases per district")
axes[0].axis("off")

# Plot 2: Cases per 1000 residents
vmin2 = zh_dist_pop_gdf["cases_per_1000"].min()
vmax2 = zh_dist_pop_gdf["cases_per_1000"].max()

zh_dist_pop_gdf.plot(
    column="cases_per_1000",
    cmap="OrRd",
    legend=True,
    vmin=vmin2,
    vmax=vmax2,
    edgecolor="black",
    linewidth=0.5,
    ax=axes[1]
)

for idx, row in zh_dist_pop_gdf.iterrows():
    point = row["geometry_poly"].representative_point()
    axes[1].annotate(
        text=row["qname"],
        xy=(point.x, point.y),
        ha="center", va="center",
        fontsize=8, color="black"
    )

axes[1].set_title("Cases per 1000 residents")
axes[1].axis("off")

plt.tight_layout()
fig.savefig("outputs/zwn_cases_per_district_total_1000_residents.png", bbox_inches="tight")
plt.show()
```

### **2.7 Visulaize the populations density per district against the cases per 1000 residents.**

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 8))

ax.scatter(
    zh_dist_pop_gdf["pop_mean_density"],
    zh_dist_pop_gdf["cases_per_1000"],
    s=100,
    color="steelblue",
    alpha=0.7,
    edgecolors="white",
    linewidth=0.5
)

# Label the district names
for idx, row in zh_dist_pop_gdf.iterrows():
    ax.annotate(
        text=row["qname"],
        xy=(row["pop_mean_density"], row["cases_per_1000"]),
        xytext=(4, 0),
        textcoords="offset points",
        fontsize=7,
        color="black"
    )

ax.set_xlabel("Population density [population / km²]")
ax.set_ylabel("Cases per 1000 residents")
ax.set_title("Population density vs. cases per 1000 residents")
ax.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
fig.savefig("outputs/zwn_cases_per_district_pop_density_1000_residents.png", bbox_inches="tight")
plt.show()
```

### **2.8 Creat a corsstable to see how often a service_code appears in a district.**

```python
ct_zwn_zh_dist_pop_gdf = pd.crosstab(zwn_zh_dist_pop_gdf["qname"], zwn_zh_dist_pop_gdf["service_code"])

print(ct_zwn_zh_dist_pop_gdf)
```

### **2.9 Creat a Heatmap to see how often a service_code appears in a district.**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Prepare data for Plot 2
counts = (
    zwn_zh_dist_pop_gdf
    .groupby(["qname", "service_code"])
    .size()
    .reset_index(name="count")
)

pop_mean_lookup = (
    zwn_zh_dist_pop_gdf[["qname", "pop_mean"]]
    .drop_duplicates("qname")
)
counts = counts.merge(pop_mean_lookup, on="qname", how="left")
counts["per_1000"] = (counts["count"] / counts["pop_mean"]) * 1000

heatmap_data = counts.pivot(
    index="qname",
    columns="service_code",
    values="per_1000"
).fillna(0)

# Plot side by side
fig, axes = plt.subplots(1, 2, figsize=(24, 10))

# Plot 1: Absolute Frequency
sns.heatmap(
    ct_zwn_zh_dist_pop_gdf,
    cmap="Reds",
    linewidths=0.5,
    annot=False,
    cbar_kws={"label": "Frequency"},
    ax=axes[0]
)
axes[0].set_title("Frequency of Service Codes per district")
axes[0].set_xlabel("Service Code")
axes[0].set_ylabel("District")

# Plot 2: Per 1000 inhabitants
sns.heatmap(
    heatmap_data,
    cmap="Reds",
    linewidths=0.5,
    cbar_kws={"label": "Cases per 1000 inhabitants"},
    ax=axes[1]
)
axes[1].set_title("Service Codes per 1000 inhabitants by District")
axes[1].set_xlabel("Service Code")
axes[1].set_ylabel("District")

plt.tight_layout()
fig.savefig("outputs/zwn_service_codes_per_district_frequency_1000_residents.png", bbox_inches="tight")
plt.show()
```

### **2.10 Filter for reports on e-scooters.**

```python
etrotti_gdf = zwn_zh_dist_pop_gdf.copy()

# Combine text columns
text_cols = ["service_code", "service_name", "title", "detail", "description"]
etrotti_gdf["text_all"] = etrotti_gdf[text_cols].fillna("").agg(" ".join, axis=1).str.lower()

# Filter e-scooters
pattern = r"e[-\s]?trott|trottinett|trotti|e[-\s]?scooter|escooter|leihroller|roller"

etrotti_gdf = etrotti_gdf[etrotti_gdf["text_all"].str.contains(pattern, regex=True, na=False)].copy()

print(len(etrotti_gdf))
etrotti_gdf.head()
etrotti_gdf.info()
```

### **2.11 Plot the reports on e-scooters acrosss the city of Zurich.**

```python
fig, ax = plt.subplots(figsize=(10, 10))

# Plot e-scooter reports
etrotti_gdf.plot(
    ax=ax,
    markersize=8,
    alpha=0.5
)

# Plot district polygons
zh_dist_pop_gdf.plot(
    ax=ax,
    color="white",
    edgecolor="black",
    alpha=0.4
)

# Add distric names
for idx, row in zh_dist_pop_gdf.iterrows():
    point = row["geometry_poly"].representative_point()

    ax.annotate(
        text=row["qname"],
        xy=(point.x, point.y),
        ha="center",
        fontsize=8,
        color="black"
    )

plt.title("Reports about e-scooters in Zurich")
plt.axis("off")
fig.savefig("outputs/test.png", bbox_inches="tight")
plt.show()
```

### **2.12 Identify the number of e-scooter reports per district.**

```python
etrotti_per_district_gdf = (
    etrotti_gdf
    .groupby("qname")
    .size()
    .reset_index(name="etrotti_count")
)
display(etrotti_per_district_gdf)
```

### **2.13 Visualize the number of e-scooter reports across the city over time.**

```python
import matplotlib.dates as mdates

# Assuming your DataFrame is called `df` and has a date column like `date` or `report_date`
etrotti_gdf["year_month"] = pd.to_datetime(etrotti_gdf["requested_date"]).dt.to_period("M").dt.to_timestamp()

cases_per_month = (
    etrotti_gdf.groupby("year_month")
    .size()
    .reset_index(name="n_cases")
)

# COVID-19 pandemic
corona_start = pd.Timestamp("2020-03-01")
corona_end   = pd.Timestamp("2022-04-30")

# Plot
fig, ax = plt.subplots(figsize=(12, 5))

# Highlight the COVID-19 pandemic
ax.axvspan(corona_start, corona_end,
           color="red", alpha=0.12, label="COVID-19 pandemic", zorder=1)
ax.axvline(corona_start, color="red", linewidth=1, linestyle="--", alpha=0.6, zorder=2)
ax.axvline(corona_end,   color="red", linewidth=1, linestyle="--", alpha=0.6, zorder=2)

# Label COVID-19 pandemic
y_max = cases_per_month["n_cases"].max()
ax.text(
    corona_start + (corona_end - corona_start) / 2,
    y_max * 0.97,
    "COVID-19 pandemic",
    ha="center", va="top", fontsize=9,
    color="red", fontweight="bold"
)

ax.plot(
    cases_per_month["year_month"],
    cases_per_month["n_cases"],
    color="#2563EB",
    linewidth=2,
    marker="o",
    markersize=4,
    markerfacecolor="white",
    markeredgewidth=1.5,
    zorder=3,
    label="Number of reports"
)

# Achsen formatieren
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b %Y"))
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
plt.xticks(rotation=45, ha="right")

ax.set_xlabel("Date", fontsize=11)
ax.set_ylabel("Number of reports", fontsize=11)
ax.set_title("Monthly electric scooter reports", fontsize=14, fontweight="bold")
ax.legend(fontsize=9)
ax.grid(axis="y", linestyle="--", alpha=0.5)
ax.spines[["top", "right"]].set_visible(False)

plt.tight_layout()
fig.savefig("outputs/zwn_electric_scooter_reports.png", bbox_inches="tight")
plt.show()
```

### **2.16 Visualize the e-scooter reports per district in absolut numbers and per 1000 residents**

```python
import matplotlib.pyplot as plt

# Fälle pro 1000 Einwohner berechnen
zh_dist_pop_gdf["etrotti_per_1000"] = (
    zh_dist_pop_gdf["etrotti_count"] / zh_dist_pop_gdf["pop_mean"]
) * 1000

fig, axes = plt.subplots(1, 2, figsize=(24, 12))

# --- Plot 1: Gesamtfälle ---
vmin1 = zh_dist_pop_gdf["etrotti_count"].min()
vmax1 = zh_dist_pop_gdf["etrotti_count"].max()

zh_dist_pop_gdf.plot(
    column="etrotti_count",
    cmap="OrRd",
    legend=True,
    vmin=vmin1,
    vmax=vmax1,
    edgecolor="black",
    linewidth=0.5,
    ax=axes[0]
)

for idx, row in zh_dist_pop_gdf.iterrows():
    point = row["geometry_poly"].representative_point()
    axes[0].annotate(
        text=row["qname"],
        xy=(point.x, point.y),
        ha="center", va="center",
        fontsize=8, color="black"
    )

axes[0].set_title("Total number of e-scooter cases per district")
axes[0].axis("off")

# --- Plot 2: Fälle pro 1000 Einwohner ---
vmin2 = zh_dist_pop_gdf["etrotti_per_1000"].min()
vmax2 = zh_dist_pop_gdf["etrotti_per_1000"].max()

zh_dist_pop_gdf.plot(
    column="etrotti_per_1000",
    cmap="OrRd",
    legend=True,
    vmin=vmin2,
    vmax=vmax2,
    edgecolor="black",
    linewidth=0.5,
    ax=axes[1]
)

for idx, row in zh_dist_pop_gdf.iterrows():
    point = row["geometry_poly"].representative_point()
    axes[1].annotate(
        text=row["qname"],
        xy=(point.x, point.y),
        ha="center", va="center",
        fontsize=8, color="black"
    )

axes[1].set_title("E-scooter cases per 1000 residents")
axes[1].axis("off")

plt.tight_layout()
fig.savefig("outputs/zwn_electric_scooter_reports_per_district_total_1000_residents.png", bbox_inches="tight")
plt.show()
```
