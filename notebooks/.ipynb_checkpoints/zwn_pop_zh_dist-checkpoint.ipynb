**Importing pandas and geopandas.**

**Loading the csv. Züri wie neu reports as pandas dataframe.**

**Inspecting the metadata and data.**

```python
import pandas as pd
import geopandas as gpd

# 1. Load the csv
zwn_df = pd.read_csv("data/raw/stzh.zwn_meldungen_p.csv")

print("Data loaded successfully!")

# 2. Peek into the data
display(zwn_df.head(4))
display(zwn_df.tail(4))

# 3. View shape & columns of the pd dataframe
print(zwn_df.shape)
print(zwn_df.columns.values)
display(zwn_df.info())
```

**Conversion of datetime data save as Dtype str to pd datetime objects.**

```python
# Conversion to pd datetime objects
for col in zwn_df.columns:
    if col.endswith("_datetime"):
        zwn_df[col] = pd.to_datetime(
            zwn_df[col],
            errors="coerce"
        )

display(zwn_df.info())
```

**Save dataframe of the Züri wie neu reports to a new csv file.**

```python
# Use index=False to prevent saving the sequential row numbers
zwn_df.to_csv("data/processed/stzh.zwn_meldungen_p_datetime64.csv", index=False)

print("File successfully saved!")
```

**Filling missing date in the column agency_sent_datetime with interpolation.**

```python
# 1. Make a safe copy
zwn_interp = zwn_df.copy()

# 2. Use interpolate() between the day before and day after to calculate the missing values in the column agency_sent_datetime
zwn_interp["agency_sent_datetime"] = zwn_interp["agency_sent_datetime"].interpolate()

display(zwn_interp.info())
```

**Normalize the datetime objects becaus only the date and not the exact time is relevant for the analysis.**

```python
#Normalization of the datetime objects and save it as new columns
for col in zwn_interp.columns:
    if col.endswith("_datetime"):

        new_col = col.replace("_datetime", "_date")

        zwn_interp[new_col] = (
            zwn_interp[col]
            .dt.normalize()
        )

display(zwn_interp["requested_date"].min())
display(zwn_interp.info())
display(zwn_interp.head(4))
```

**Rename the geometry column.**

```python
# Rename geometry column
zwn_interp = zwn_interp.rename(
    columns={"geometry": "geometry_zwn"}
)
# View change
display(zwn_interp.info())
```

**Save dataframe of the Züri wie neu reports to a new csv file.**

```python
# Use index=False to prevent saving the sequential row numbers
zwn_interp.to_csv("data/processed/stzh.zwn_meldungen_p_datetime64_normalized.csv", index=False)

print("File successfully saved!")
```

**Create a geo dataframe of the Züri wie neu reports.**

```python
from shapely import wkt

# WKT -> Geometrie konvertieren
zwn_interp["geometry_zwn"] = zwn_interp["geometry_zwn"].apply(wkt.loads)

# GeoDataFrame erstellen
zwn_interp_gdf = gpd.GeoDataFrame(zwn_interp, geometry="geometry_zwn")

# CRS setzen (LV95 / EPSG:2056)
zwn_interp_gdf = zwn_interp_gdf.set_crs(epsg=2056)

# 2. Peek into the data an inspect the coordinate reference system
print(zwn_interp_gdf.crs)
display(zwn_interp_gdf.head(4))

# 3. View shape & columns of the geo dataframe
print(zwn_interp_gdf.shape)
print(zwn_interp_gdf.columns.values)
print(zwn_interp_gdf.crs)
display(zwn_interp_gdf.info())
```

**Save geo dataframe of the Züri wie neu reports to a new gpkg file.**

```python
# Export as a GeoPackage (for sharing with QGIS/ArcGIS users)
zwn_interp_gdf.to_file("data/processed/stzh.zwn_meldungen_p_processed.gpkg")

print("File successfully saved!")
```

**Loading the xlsx. Bevoelkerung and convert it to a pandas dataframe.**

**Inspecting the metadata and data.**

```python
# 1. Load the xlsx
file_path = "data/raw/Bevoelkerung_nach-Herkunft-Stadtkreis-Stadtquartier.xlsx"

# Years / Sheets
years = [str(year) for year in range(2025, 2013, -1)]

dfs = []

for year in years:

    # Excel einlesen
    df = pd.read_excel(
        file_path,
        sheet_name=year,
        usecols="A:B",
        skiprows=8,
        header=0,
        engine="openpyxl"
    )

    # Spalten umbenennen
    df.columns = ["qname", f"poptot_{year}"]

    # Optional: Leere Zeilen entfernen
    df = df.dropna(subset=["qname"])

    dfs.append(df)

# Alle Jahre über districts zusammenführen
zh_pop_df = dfs[0]

for df in dfs[1:]:
    zh_pop_df = zh_pop_df.merge(df, on="qname", how="outer")

print("Data loaded successfully!")

# 2. Peek into the data
display(zh_pop_df.head(4))
display(zh_pop_df.tail(4))

# 3. View shape & columns of the pd dataframe
print(zh_pop_df.shape)
print(zh_pop_df.columns.values)
display(zh_pop_df.info())
```

**Inspect the districts (qname) column.**

```python
zh_pop_df["qname"]
```

**Save dataframe of the Züri wie neu reports to a new csv file.**

```python
# We use index=False to prevent saving the sequential row numbers
zh_pop_df.to_csv("data/processed/Bevoelkerung_nach-Herkunft-Stadtkreis-Stadtquartier_cleaned.csv", index=False)

print("File successfully saved!")
```

**Remove the rows with  Kreis and  Ganze Stadt from the dataframe as this rows are not needed.**

```python
zh_pop_urbandists = zh_pop_df.copy()

zh_pop_urbandists = zh_pop_urbandists[
    (~zh_pop_urbandists["qname"].str.startswith("Kreis")) &
    (zh_pop_urbandists["qname"] != "Ganze Stadt")
]

zh_pop_urbandists = zh_pop_urbandists.reset_index(drop=True)

# 2. Peek into the data
display(zh_pop_urbandists.head(4))

# 3. View shape & columns of the pd dataframe
print(zh_pop_urbandists.shape)
print(zh_pop_urbandists.columns.values)
display(zh_pop_urbandists.info())

zh_pop_urbandists["qname"]
```

**Merge the geo dataframe districts population with the geo dataframe quartiere.**

```python
# Merg the two geo dataframes
zh_dist_pop_gdf= zh_dist_gdf.merge(zh_pop_urbandists, on="qname", how="outer")

print("Merge successfully!")

# 2. Peek into the data
print(zh_dist_pop_gdf.crs)
display(zh_dist_pop_gdf.head(4))

# 3. View shape & columns of the geo dataframe
print(zh_dist_pop_gdf.shape)
print(zh_dist_pop_gdf.columns.values)
display(zh_dist_pop_gdf.info())
```

**Calculate the  area for each quartier.**

```python
zh_dist_pop_gdf["area_m2"] = zh_dist_pop_gdf.geometry.area
zh_dist_pop_gdf["area_km2"] = (
    zh_dist_pop_gdf["area_m2"] / 1_000_000
).round(2)

display(zh_dist_pop_gdf.head(4))
```

**Calculate the population denstiy of each quartier and year.**

```python
for year in range(2014, 2026):

    pop_col = f"poptot_{year}"
    density_col = f"pop_density_{year}"

    zh_dist_pop_gdf[density_col] = (
        zh_dist_pop_gdf[pop_col] /
        zh_dist_pop_gdf["area_km2"]
    ).round(2)
    
display(zh_dist_pop_gdf.head(4))
```

**Plot the population density.**

```python
# Instantly visualize the normalized density ratio
zh_dist_pop_gdf.plot(
    column="pop_density_2025",
    cmap="YlOrRd",
    figsize=(8, 5),
    legend=True,
    legend_kwds={"label": "Population Density (Inhabitants / Area)"},
    edgecolor="grey",
);
```

**Identify the number of cases per district.**

```python
cases_per_district = (
    zwn_zh_dist_pop_gdf
    .groupby("qname")
    .size()
    .reset_index(name="case_count")
)
display(zwn_zh_dist_pop_gdf.info())
display(cases_per_district)
```

**Visualize the numbers of cases per district.**

```python
zh_dist_pop_gdf.sort_values(
    by="case_count",
    ascending=False
).plot(
    x="qname",
    y="case_count",
    kind="bar",
    color="blue",
    figsize=(12, 6)
)

zh_dist_pop_gdf.plot(
    column="case_count",
    legend=True,
    cmap="OrRd"
)
```

**Merge the cases per district with the district polygon geo dateframe.**

```python
zh_dist_pop_gdf = zh_dist_pop_gdf.merge(
    cases_per_district,
    on="qname",  # Spaltenname anpassen
    how="left"   # left = alle Quartiere bleiben erhalten, auch leere
)
zh_dist_pop_gdf["case_count"] = zh_dist_pop_gdf["case_count"].fillna(0)

display(zh_dist_pop_gdf.info())

display(zh_dist_pop_gdf.head(4))
```

**Merge the cases per district per year with the district polygon geo dateframe.**

```python
zwn_zh_dist_pop_gdf['year'] = zwn_zh_dist_pop_gdf['requested_date'].dt.year

display(zwn_zh_dist_pop_gdf.info())

cases_per_district_year = (
    zwn_zh_dist_pop_gdf[zwn_zh_dist_pop_gdf['year'].between(2014, 2025)]
    .groupby(["qname", "year"])
    .size()
    .reset_index(name="case_count")
)



# Wide-Format: eine Spalte pro Jahr
cases_wide = cases_per_district_year.pivot(
    index="qname",
    columns="year",
    values="case_count"
).fillna(0).astype(int).reset_index()

# Spalten umbenennen: 2014 → case_count_2014
cases_wide.columns = [
    f"case_count_{col}" if isinstance(col, int) else col
    for col in cases_wide.columns
]
cases_wide.columns.name = None

# Merge mit dem Geo-DataFrame
zh_dist_pop_gdf = zh_dist_pop_gdf.merge(cases_wide, on="qname", how="outer")

print("Merge erfolgreich!")
display(zh_dist_pop_gdf.info())
display(zh_dist_pop_gdf.head(4))
```

**Visualize the cases per district for distinct years.**

```python
import matplotlib.pyplot as plt

# DataFrame
df = zh_dist_pop_gdf.copy()

# Nur gewünschte Jahre auswählen
year_cols = [
    "case_count_2014",
    "case_count_2017",
    "case_count_2021",
    "case_count_2025"
]

# DataFrame von wide -> long umformen
df_long = df.melt(
    id_vars="qname",
    value_vars=year_cols,
    var_name="Jahr",
    value_name="Fälle"
)

# Jahresnamen bereinigen
df_long["Jahr"] = df_long["Jahr"].str.replace("case_count_", "")

# Pivot-Tabelle erstellen
pivot_df = df_long.pivot(
    index="qname",
    columns="Jahr",
    values="Fälle"
)

# Bar Chart erstellen
pivot_df.plot(
    kind="bar",
    figsize=(16, 8)
)

plt.title("Fälle pro Jahr und Quartier")
plt.xlabel("Quartier")
plt.ylabel("Anzahl Fälle")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Jahr")

plt.tight_layout()
plt.show()
```

**Visualize the cases per district for distinct years per 1000 inhabitants.**

```python
import pandas as pd
import matplotlib.pyplot as plt

# DataFrame
df = zh_dist_pop_gdf.copy()

# Gewünschte Jahre
years = [2014, 2017, 2021, 2025]

# Fälle pro 1000 Einwohner berechnen
for year in years:
    df[f"rate_{year}"] = (
        df[f"case_count_{year}"] / df[f"poptot_{year}"]
    ) * 1000

# Neue Rate-Spalten
rate_cols = [f"rate_{year}" for year in years]

# Wide -> Long
df_long = df.melt(
    id_vars="qname",
    value_vars=rate_cols,
    var_name="Jahr",
    value_name="Fälle pro 1000 Einwohner"
)

# Jahresnamen bereinigen
df_long["Jahr"] = df_long["Jahr"].str.replace("rate_", "")

# Pivot-Tabelle
pivot_df = df_long.pivot(
    index="qname",
    columns="Jahr",
    values="Fälle pro 1000 Einwohner"
)

# Bar Chart
pivot_df.plot(
    kind="bar",
    figsize=(18, 8)
)

# Layout
plt.title("Fälle pro 1000 Einwohner nach Quartier")
plt.xlabel("Quartier")
plt.ylabel("Fälle pro 1000 Einwohner")
plt.xticks(rotation=45, ha="right")

plt.legend(title="Jahr")
plt.tight_layout()

plt.show()
```

**Visulaize the populations density per district against the cases per 1000 inhabitants.**

```python
import matplotlib.pyplot as plt

# DataFrame
df = zh_dist_pop_gdf.copy()

# Gewünschte Jahre
years = [2014, 2017, 2021, 2025]

# 2x2 Subplots erstellen
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Achsen flach machen
axes = axes.flatten()

# Für jedes Jahr ein eigener Scatterplot
for i, year in enumerate(years):

    ax = axes[i]

    # Fälle pro 1000 Einwohner berechnen
    rate = (
        df[f"case_count_{year}"] / df[f"poptot_{year}"]
    ) * 1000

    # Scatterplot
    ax.scatter(
        df[f"pop_density_{year}"],
        rate,
        s=80,
        alpha=0.7
    )

    # Quartiernamen hinzufügen
    for _, row in df.iterrows():
        ax.text(
            row[f"pop_density_{year}"],
            (
                row[f"case_count_{year}"] /
                row[f"poptot_{year}"]
            ) * 1000,
            row["qname"],
            fontsize=8
        )

    # Titel und Labels
    ax.set_title(f"Jahr {year}")
    ax.set_xlabel("Bevölkerungsdichte")
    ax.set_ylabel("Fälle pro 1000 Einwohner")

    # Grid
    ax.grid(True, alpha=0.3)

# Gesamttitel
fig.suptitle(
    "Bevölkerungsdichte vs. Fälle pro 1000 Einwohner",
    fontsize=16
)

plt.tight_layout()
plt.show()
```
