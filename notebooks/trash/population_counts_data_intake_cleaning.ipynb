```python
import pandas as pd

# 1. Load, sep=";" or "," test it!
zh_pop_df = pd.read_csv("data/raw/Bevölkerungsentwicklung_Stadtkreis_Stadtquartier.csv", sep=";", encoding="utf-8")

print("Data loaded successfully!")

# 2. Peek
display(zh_pop_df.head(10))
display(zh_pop_df.tail(10))

# 3. Shape & Columnsb
print(zh_pop_df.shape)
print(zh_pop_df.columns.values)
display(zh_pop_df.info())
```

```python
for col in zh_pop_df.columns:
    print(f"{col}: {zh_pop_df[col].head(10).tolist()}")
```

```python
# Alle Spalten die als Jahreszahl erkennbar sind
year_cols = [col for col in zh_pop_df.columns if col.isdigit()]
print(year_cols)  # Kontrolle

# Alle Jahres-Spalten auf einmal konvertieren
zh_pop_df[year_cols] = zh_pop_df[year_cols].apply(
    lambda col: pd.to_numeric(col.str.replace(r"[^\d]", "", regex=True), errors="coerce")
)

# 2. Peek
display(zh_pop_df.head(10))
display(zh_pop_df.tail(10))

# 3. Shape & Columns
print(zh_pop_df.shape)
print(zh_pop_df.columns.values)
display(zh_pop_df.info())
```

```python
# Save DataFrame to a new CSV file
# We use index=False to prevent saving the sequential row numbers
zh_pop_df.to_csv("data/processed/Bevölkerungsentwicklung_Stadtkreis_Stadtquartier_cleaned.csv", index=False)

print("File successfully saved!")
```

```python
import pandas as pd
import numpy as np

zh_pop_interp = zh_pop_df.copy()

# Neue Jahres-Spalten hinzufügen
for year in range(2011, 2025):
    zh_pop_interp[str(year)] = np.nan

# Jahres-Spalten sortieren
year_cols = [str(y) for y in sorted(map(int, zh_pop_interp.columns[1:]))]

# districts + sortierte Jahres-Spalten
zh_pop_interp = zh_pop_interp[["districts"] + year_cols]

# Nur Jahresdaten interpolieren
zh_pop_interp[year_cols] = (
    zh_pop_interp[year_cols]
    .astype(float)
    .interpolate(axis=1)
    .round()
    .astype(int)
)

# 2. Peek
display(zh_pop_interp.head(10))
display(zh_pop_interp.tail(10))

# 3. Shape & Columns
print(zh_pop_interp.shape)
print(zh_pop_interp.columns.values)
display(zh_pop_interp.info())
```

```python
# Save DataFrame to a new CSV file
# We use index=False to prevent saving the sequential row numbers
zh_pop_interp.to_csv("data/processed/Bevölkerungsentwicklung_Stadtkreis_Stadtquartier_interpolated.csv", index=False)

print("File successfully saved!")
```

```python
zh_pop_urbandists = zh_pop_interp.copy()

zh_pop_urbandists = zh_pop_urbandists[
    (~zh_pop_urbandists["districts"].str.startswith("Kreis")) &
    (zh_pop_urbandists["districts"] != "Ganze Stadt")
]

zh_pop_urbandists = zh_pop_urbandists.reset_index(drop=True)

# 2. Peek
display(zh_pop_urbandists.head(10))
display(zh_pop_urbandists.tail(10))

# 3. Shape & Columns
print(zh_pop_urbandists.shape)
print(zh_pop_urbandists.columns.values)
display(zh_pop_urbandists.info())
```

```python
# Save DataFrame to a new CSV file
# We use index=False to prevent saving the sequential row numbers
zh_pop_urbandists.to_csv("data/processed/Bevölkerungsentwicklung_Stadtkreis_Stadtquartier_urbandists.csv", index=False)

print("File successfully saved!")
```
