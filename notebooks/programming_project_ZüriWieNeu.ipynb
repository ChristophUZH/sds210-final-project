### **1.2 ZueriWieNeu reports**

#### **1.2.1 Loading the csv. ZueriWieNeu reports as a Pandas dataframe and inspect the data and metadata.**

```python
import pandas as pd
import geopandas as gpd

# 1. Load the csv
zwn_df = pd.read_csv("data/raw/stzh.zwn_meldungen_p.csv")

print("Data loaded successfully!")

# 2. Peek into the data
display(zwn_df.head(4))
display(zwn_df.tail(4))

# 3. View shape & columns of the pd dataframe
print(zwn_df.shape)
print(zwn_df.columns.values)
display(zwn_df.info())
```

#### **1.2.2 Conversion of datetime data to Pandas datetime objects.**

```python
# Conversion to Pandas datetime objects
for col in zwn_df.columns:
    if col.endswith("_datetime"): 
        zwn_df[col] = pd.to_datetime(zwn_df[col], errors="coerce") # If someting is not recognised it is transformed in to NaT by errors="coerce"

display(zwn_df.info())
```

#### **1.2.3 Filling in missing dates in the column agency_sent_datetime with an interpolation.**

```python
# 1. Make a safe copy
zwn_interp_df = zwn_df.copy()

# 2. Use interpolate() between the day before and day after to calculate the missing values in the column agency_sent_datetime
zwn_interp_df["agency_sent_datetime"] = zwn_interp_df["agency_sent_datetime"].interpolate()

display(zwn_interp_df.info())
```

#### **1.2.4 Normalize the datetime objects becaus only the date and not the exact time is relevant for the analysis.**

```python
#Normalization of the datetime objects and save it as new columns
for col in zwn_interp_df.columns:
    if col.endswith("_datetime"):
        new_col = col.replace("_datetime", "_date")
        zwn_interp_df[new_col] = (zwn_interp_df[col].dt.normalize())

display(zwn_interp_df["requested_date"].min())
```

#### **1.2.5 Rename the geometry column of the zwn_interp_df.**

```python
# Rename geometry column
zwn_interp_df = zwn_interp_df.rename(columns={"geometry": "geometry_zwn"})

# View change
display(zwn_interp_df.info())
```

#### **1.2.6 Save the zwn_interp_df of the Zueri wie neu reports to a new csv file.**

```python
# Use index=False to prevent saving the sequential row numbers of the Index in the csv
zwn_interp_df.to_csv("data/processed/stzh.zwn_meldungen_p_datetime64_normalized.csv", index=False)

print("File successfully saved!")
```

#### **1.2.7 Create a geo dataframe of the Zueri wie neu reports.**

#### **1.6.5 Define a function which places the distric names on the map.**

<span style="color:red; font-weight:bold">
LLM assisted code
</span>

```python
def add_district_names(gdf, ax):
    """
    Adds district names as text annotations to a matplotlib axis.

    For each row in the GeoDataFrame, the function calculates a
    representative point inside the polygon geometry and places the
    corresponding district name (`qname`) at that position.

    Parameters
    ----------
    gdf : geopandas.GeoDataFrame
        GeoDataFrame containing the district geometries and names.
        Must include the columns:
        - ``geometry_poly`` : polygon geometry
        - ``qname`` : district name to display
    ax : matplotlib.axes.Axes
        Matplotlib axis object on which the district labels are drawn.

    Returns
    -------
    None
        The function modifies the provided matplotlib axis in place.

    Examples
    --------
    >>> fig, ax = plt.subplots()
    >>> gdf.plot(ax=ax)
    >>> add_district_names(gdf, ax)
    """
    for idx, row in gdf.iterrows():
        point = row["geometry_poly"].representative_point()

        ax.annotate(
            text=row["qname"],
            xy=(point.x, point.y),
            ha="center",
            va="center",
            fontsize=8,
            color="black"
        )
```

```python
import matplotlib.dates as mdates

# cases as a synonym for reports

# Aggregate reports by month
cases_over_time = (
    zwn_zh_dist_pop_gdf
    .groupby(pd.Grouper(key="requested_date", freq="ME"))
    .size()
    .reset_index(name="count")
    .sort_values("requested_date")
)

# COVID-19 pandemic
corona_start = pd.Timestamp("2020-03-01")
corona_end   = pd.Timestamp("2022-04-30")

# Plot
fig, ax = plt.subplots(figsize=(12, 6))

# Highlight the COVID-19 period
ax.axvspan(corona_start, corona_end,
           color="red", alpha=0.12, label="COVID-19 period", zorder=1)
ax.axvline(corona_start, color="red", linewidth=1, linestyle="--", alpha=0.6, zorder=2)
ax.axvline(corona_end,   color="red", linewidth=1, linestyle="--", alpha=0.6, zorder=2)

# Showing the COVID-19 period in the plot
y_max = cases_over_time["count"].max()

ax.text(
    corona_start + (corona_end - corona_start) / 2,
    y_max * 0.90,
    "COVID-19 period",
    ha="center", va="top", fontsize=9,
    color="red", fontweight="bold"
)

ax.plot(
    cases_over_time["requested_date"],
    cases_over_time["count"],
    color="blue",
    linewidth=2,
    marker="o",
    markersize=4,
    markerfacecolor="white",
    markeredgewidth=1.5,
    zorder=3,
    label="Number of reports"
)

# Formating the axes
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b %Y"))
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
plt.xticks(rotation=45, ha="right")
ax.set_xlabel("Date", fontsize=11)
ax.set_ylabel("Number of reports", fontsize=11)
ax.set_title("ZueriWieNeu reports over time", fontsize=14, fontweight="bold")
ax.legend(fontsize=9)
ax.grid(axis="y", linestyle="--", alpha=0.5)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
fig.savefig("outputs/zwn_reports_over_time.png", bbox_inches="tight")
plt.show()
```

### **2.5 Define a function which calculates the the reports per 1000 residents.**

<span style="color:red; font-weight:bold">
LLM assisted code
</span>

```python
def calculate_per_1000(gdf, count_col, population_col="pop_mean"):
    """
    Calculates the reports per 1000 inhabitants for a given count column.

    The function divides the values in ``count_col`` by the corresponding
    population values in ``population_col`` and multiplies the result by 1000.

    Parameters
    ----------
    gdf : geopandas.GeoDataFrame
        GeoDataFrame containing the count and population data.
    count_col : str
        Name of the column containing the counts or events of interest.
    population_col : str, optional
        Name of the column containing population values.
        Default is ``"pop_mean"``.

    Returns
    -------
    geopandas.Series
        A Series containing the reports per 1,000 inhabitants.

    Examples
    --------
    >>> calculate_per_1000(gdf, "cases")
    0    12.4
    1     8.7
    2    15.1
    dtype: float64
    """
    return (gdf[count_col] / gdf[population_col]) * 1000
```
